```python
def calculate_sharpness(image):
    """
    Higher value = generally sharper image.
    Lower value = generally blurrier image.
    """

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_RGB2GRAY
    )

    return cv2.Laplacian(
        gray,
        cv2.CV_64F
    ).var()
```

```python
quality_data = []

for filename in image_files:

    path = os.path.join(IMAGE_DIR, filename)

    image = cv2.imread(path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    sharpness = calculate_sharpness(image)

    quality_data.append({
        "image": filename,
        "sharpness": sharpness
    })

for row in quality_data:
    print(
        row["image"],
        "→ Sharpness:",
        round(row["sharpness"], 2)
    )
```

```python
quality_data_sorted = sorted(
    quality_data,
    key=lambda x: x["sharpness"]
)

print("Potentially blurrier images:\n")

for row in quality_data_sorted:
    print(
        row["image"],
        "→",
        round(row["sharpness"], 2)
    )
```

```python
import re
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

def get_image_id(filename):
    """
    Extract the numeric patient/image ID from the filename.
    Example:
    01_training.tif -> 01
    01_training_mask.gif -> 01
    """
    match = re.search(r'\d+', filename)
    return match.group(0) if match else None


image_map = {}
mask_map = {}

for f in image_files:
    image_id = get_image_id(f)
    if image_id:
        image_map[image_id] = f

for f in mask_files:
    mask_id = get_image_id(f)
    if mask_id:
        mask_map[mask_id] = f


print("Image IDs:", sorted(image_map.keys()))
print("Mask IDs :", sorted(mask_map.keys()))

common_ids = sorted(set(image_map.keys()) & set(mask_map.keys()))

print("\nMatched pairs:", len(common_ids))

for image_id in common_ids:
    print(
        image_map[image_id],
        "<--->",
        mask_map[image_id]
    )
```

```python
fig, axes = plt.subplots(4, 5, figsize=(20, 16))

for ax, image_id in zip(axes.ravel(), common_ids):

    image_name = image_map[image_id]
    mask_name = mask_map[image_id]

    image = cv2.imread(
        os.path.join(IMAGE_DIR, image_name)
    )

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    mask = cv2.imread(
        os.path.join(MASK_DIR, mask_name),
        cv2.IMREAD_GRAYSCALE
    )

    ax.imshow(image)
    ax.imshow(mask, cmap="gray", alpha=0.20)

    ax.set_title(
        f"ID {image_id}"
    )

    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
fov_data = []

for image_id in common_ids:

    mask_name = mask_map[image_id]

    mask = cv2.imread(
        os.path.join(MASK_DIR, mask_name),
        cv2.IMREAD_GRAYSCALE
    )

    # Binary FOV mask
    fov = mask > 0

    total_pixels = fov.size
    retinal_pixels = np.sum(fov)

    fov_percentage = (
        retinal_pixels / total_pixels
    ) * 100

    fov_data.append({
        "image": image_map[image_id],
        "id": image_id,
        "fov_percentage": fov_percentage
    })

for row in fov_data:
    print(
        row["image"],
        "→ FOV:",
        round(row["fov_percentage"], 2),
        "%"
    )
```

```python
quality_results = []

for image_id in common_ids:

    image_name = image_map[image_id]
    mask_name = mask_map[image_id]

    # -------------------------
    # Load image
    # -------------------------
    image = cv2.imread(
        os.path.join(IMAGE_DIR, image_name)
    )

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    # -------------------------
    # Load FOV mask
    # -------------------------
    mask = cv2.imread(
        os.path.join(MASK_DIR, mask_name),
        cv2.IMREAD_GRAYSCALE
    )

    fov = mask > 0

    # -------------------------
    # FOV coverage
    # -------------------------
    fov_percentage = (
        np.sum(fov) / fov.size
    ) * 100

    # -------------------------
    # Pixels inside retina only
    # -------------------------
    retinal_pixels = gray[fov]

    # -------------------------
    # Brightness
    # -------------------------
    brightness = np.mean(retinal_pixels)

    # -------------------------
    # Contrast
    # -------------------------
    contrast = np.std(retinal_pixels)

    # -------------------------
    # Sharpness
    # -------------------------
    laplacian = cv2.Laplacian(
        gray,
        cv2.CV_64F
    )

    sharpness = np.var(
        laplacian[fov]
    )

    # -------------------------
    # Dark pixel percentage
    # -------------------------
    dark_percentage = (
        np.sum(retinal_pixels < 30)
        / len(retinal_pixels)
    ) * 100

    # -------------------------
    # Very bright pixel percentage
    # -------------------------
    bright_percentage = (
        np.sum(retinal_pixels > 245)
        / len(retinal_pixels)
    ) * 100

    quality_results.append({
        "image": image_name,
        "id": image_id,
        "sharpness": sharpness,
        "brightness": brightness,
        "contrast": contrast,
        "fov_percentage": fov_percentage,
        "dark_percentage": dark_percentage,
        "bright_percentage": bright_percentage
    })


quality_df = pd.DataFrame(
    quality_results
)

quality_df
```

```python
def min_max_normalize(series):
    min_value = series.min()
    max_value = series.max()

    if max_value == min_value:
        return pd.Series(
            np.ones(len(series)),
            index=series.index
        )

    return (
        series - min_value
    ) / (
        max_value - min_value
    )


quality_df["sharpness_score"] = min_max_normalize(
    quality_df["sharpness"]
)

quality_df["brightness_score"] = min_max_normalize(
    quality_df["brightness"]
)

quality_df["contrast_score"] = min_max_normalize(
    quality_df["contrast"]
)

quality_df["fov_score"] = min_max_normalize(
    quality_df["fov_percentage"]
)
```

```python
quality_df["preliminary_quality_score"] = (
    0.40 * quality_df["sharpness_score"] +
    0.25 * quality_df["contrast_score"] +
    0.20 * quality_df["fov_score"] +
    0.15 * quality_df["brightness_score"]
)

quality_df = quality_df.sort_values(
    "preliminary_quality_score"
)

display(
    quality_df[[
        "image",
        "sharpness",
        "brightness",
        "contrast",
        "fov_percentage",
        "preliminary_quality_score"
    ]]
)
```
